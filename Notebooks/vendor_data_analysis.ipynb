```python
# Creating freight Summary Table

query = """
DROP TABLE IF EXISTS freight_summary;

CREATE TABLE freight_summary AS
SELECT
    VendorNumber,
    VendorName,
    SUM(Freight) AS TotalFreight
FROM vendor_invoice
GROUP BY VendorNumber, VendorName;
"""

conn.executescript(query)
conn.commit()

# Verifying the table

freight_summary = pd.read_sql_query(
    "SELECT * FROM freight_summary LIMIT 5",
    conn
)

freight_summary
```

```python
# Creating Purchase summary table
query="""
drop table if exists purchase_summary;

create table purchase_summary as 
select
    p.VendorNumber,
    p.VendorName,
    p.Brand,
    pp.Description,
    sum(p.Quantity) as total_pur_qty,
    sum(p.Dollars) as total_pur_dollars,
    avg(pp.Price) as avg_retail_price,
    avg(pp.PurchasePrice) as avg_purchase_price
from purchases p
join purchase_prices pp
    on p.Brand=pp.Brand
group by 
    p.VendorNumber,
    p.VendorName,
    p.Brand,
    pp.Description;
"""
conn.executescript(query)
conn.commit()

# Verifying the table
purchase_summary=pd.read_sql_query(
    "select * from purchase_summary limit 5",conn
)

purchase_summary
```

```python
# Sales table summary
query = """
DROP TABLE IF EXISTS sales_summary;

CREATE TABLE sales_summary AS
SELECT
    VendorNo AS VendorNumber,
    VendorName,
    Brand,
    Description,
    SUM(SalesQuantity) AS TotalSalesQuantity,
    SUM(SalesDollars) AS TotalSalesDollars
FROM sales
GROUP BY
    VendorNo,
    VendorName,
    Brand,
    Description;
"""

conn.executescript(query)
conn.commit()

# Verifying the table
sales_summary = pd.read_sql_query(
    "SELECT * FROM sales_summary LIMIT 5",
    conn
)

sales_summary
```

```python
# Checking columns in each summary table

for table in ["purchase_summary","sales_summary","freight_summary"]:
    print(f"\n{table}")
    display(pd.read_sql_query(
        f"select * from {table} limit 3",conn
    ))
```

```python
# Combining all the table summary as vendor_sales_summary

query="""
drop table if exists vendor_sales_summary;

create table vendor_sales_summary as 
select
    ps.VendorNumber,
    ps.VendorName,
    ps.Brand,
    ps.Description,
    ps.total_pur_qty,
    ps.total_pur_dollars,
    ps.avg_retail_price,
    ps.avg_purchase_price,
    ss.TotalSalesQuantity,
    ss.TotalSalesDollars,
    fs.TotalFreight
from purchase_summary ps
left join sales_summary ss
    on ps.VendorNumber=ss.VendorNumber
    and ps.Brand=ss.Brand
left join freight_summary fs
    on ps.VendorNumber=fs.VendorNumber
"""

conn.executescript(query)
conn.commit()

#verifying the table
pd.read_sql_query(
    "SELECT * FROM vendor_sales_summary LIMIT 5",
    conn
)
```

```python
# Calculate the Gross Profit

vendor_sales_summary["GrossProfit"] = (
    vendor_sales_summary["TotalSalesDollars"]
    - vendor_sales_summary["total_pur_dollars"]
)

# Calculate Profit Margin

vendor_sales_summary["ProfitMargin"] = (
    vendor_sales_summary["GrossProfit"]
    / vendor_sales_summary["TotalSalesDollars"]
) * 100

# Calculate Stock TurnOver

vendor_sales_summary["StockTurnover"] = (
    vendor_sales_summary["TotalSalesQuantity"]
    / vendor_sales_summary["total_pur_qty"]
)

# Calculate Sales-to-Purchase Ratio

vendor_sales_summary["SalesToPurchaseRatio"] = (
    vendor_sales_summary["TotalSalesDollars"]
    / vendor_sales_summary["total_pur_dollars"]
)

# Save the final Table
vendor_sales_summary.to_sql(
    "vendor_sales_summary",
    conn,
    if_exists="replace",
    index=False
)

pd.read_sql_query(
    "select * from vendor_sales_summary limit 5",
    conn
)
```

```python
# Database Connection
import pandas as pd
import os
import logging
import time
from sqlalchemy import create_engine

data="../Data"
engine=create_engine("sqlite:///../inventory.db")

# Logging

import logging
os.makedirs("../Logs",exist_ok=True)

logging.basicConfig(
    filename="../Logs/ingestion_db.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode="a"
)

logging.info("Logging started")


# Ingestion function

def ingest_db(file_path,table_name,engine):
    try:
        
        first_chunk=True
        for chunk in pd.read_csv(file_path,chunksize=100000):
            chunk.to_sql(
                table_name,
                con=engine,
                if_exists="replace" if first_chunk else "append",
                index=False
            )
            first_chunk=False
    
        logging.info(f"{table_name} imported successfully")
    except Exception as e:
        logging.error(f"Error importing {table_name}: {e}")

# Import all csv files

def load_raw_data():
    
    for file in os.listdir(data):
        if file.endswith(".csv"):
            file_path=os.path.join(data,file)
            table_name=file[:-4]
            ingest_db(file_path,table_name,engine)
            print(f"{table_name} imported successfully!")

# scripting

if __name__ == "__main__":
    start_time=time.time()
    load_raw_data()
    end_time=time.time()
    total_time=end_time-start_time
    logging.info(f"Total execution time: {total_time} seconds")
    print(f"Total execution time : {total_time:.2f} seconds")
```
